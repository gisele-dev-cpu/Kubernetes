```python
# WAP to calculate average of two numbers.

number_1 = float(input("Enter the first number: "))
number_2 = float(input("Enter the second number: "))

average = (number_1 + number_2) / 2

print(f"The averager of {number_1} + {number_2} is  {average}")
```

```python
#To calculate the total amount of money in piggybank, given the coins of Rs.10, Rs5, Rs2 and Rs1.
coins_10 = int(input("Enter number of Rs.10 coin: "))
coins_5 = int(input("Enter number of Rs.5 coin: "))
coins_2 = int(input("Enter number of Rs.2 coin: "))
coins_1 = int(input("Enter number of Rs.1 coin: "))

total = (coins_10 * 10) + (coins_5 * 5) + (coins_2 * 2) + (coins_1 * 1)

print(f"\nTotal amount in the piggybank: Rs.{total}")
```

```python
# To calculate the bill amount for an item given its quality sold, value, discount and tax.

quantity = int(input("Enter quantity sold: "))
price = float(input("Enter price per item: Rs."))
discount = float(input("Enter discount (%): "))
tax = float(input("Enter tax (%): "))

# Step 1: Calculate amount before discount and tax
amount = quantity * price

# Step 2: Apply discount
discount_amount = (discount / 100) * amount
amount_after_discount = amount - discount_amount

# Step 3: Apply tax

tax_amount = (tax / 100) * amount_after_discount
final_amount = amount_after_discount + tax_amount

print(f"\nAmount before discount: Rs.{amount:.2f}")
print(f"Discount amount: Rs.{discount_amount:.2f}")
print(f"Amount after discount: Rs.{amount_after_discount:.2f}")
print(f"Tax amount: Rs.{tax_amount:.2f}")
print(f"Final bill amount: Rs.{final_amount:.2f}")
```
